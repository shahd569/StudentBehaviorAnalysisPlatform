```python
import numpy as np

features = [
    "day_1",
    "day_2",
    "day_3",
    "day_4",
    "day_5",
    "day_6",
    "day_7"
]


def explain_prediction(model, input_data):

    # 1. التنبؤ
    prediction = model.predict(input_data)[0]
    proba = model.predict_proba(input_data)[0]

    # 2. أهمية كل يوم
    importance = model.feature_importances_

    explanation = []

    # 3. تفسير حسب الأهمية
    sorted_idx = np.argsort(importance)[::-1]

    for i in sorted_idx:
        if importance[i] > 0.05:
            explanation.append(
                f"{features[i]} influenced the decision strongly"
            )

    # 4. تفسير حسب النتيجة
    if prediction == 0:
        status = "Good"
    elif prediction == 1:
        status = "Average"
    else:
        status = "At Risk"

    # 5. تفسير ذكي إضافي (مهم جداً)
    last_days = input_data[0]

    if last_days[-1] > last_days[-2]:
        explanation.append("Student performance is declining recently")

    if sum(last_days) > 10:
        explanation.append("Overall engagement is low")

    return {
        "prediction": status,
        "confidence": round(max(proba), 2),
        "reasons": explanation
    }
```
